# Distributions of variables

## temperature time series

```python
ts = df.groupby("valid_time")["t2m"].mean()

plt.figure(figsize=(14, 5))
ts.plot()
plt.title("Mean temperature per month over time")
plt.xlabel("Date")
plt.ylabel("t2m")
plt.show()
```

```python
fig, axes = plt.subplots(3, 4, figsize=(16, 9))
for ax, c in zip(axes.ravel(), FEATURES + ["t2m"]):
    sns.histplot(df[c], kde=True, ax=ax)
    ax.set_title(f"{c} (skew={df[c].skew():.1f})")
plt.tight_layout(); plt.show()
```

## Exploration 2: Outliers

### outlier counts

```python
rows = []
for c in FEATURES + ["t2m"]:
    s = df[c].dropna()
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    n = ((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum()
    rows.append({"variable": c, "outliers": n, "percent": n / len(s) * 100})

out = pd.DataFrame(rows)
print(out.round(2))

plt.figure(figsize=(10, 4))
sns.barplot(data=out, x="variable", y="percent")
plt.title("IQR outliers (%) by variable")
plt.show()
```

```python
fig, axes = plt.subplots(3, 4, figsize=(16, 9))
for ax, c in zip(axes.ravel(), FEATURES + ["t2m"]):
    sns.boxplot(y=df[c], ax=ax, fliersize=1); ax.set_title(c)
plt.tight_layout(); plt.show()
```

## SLHF vs evaporation

```python
d = df[["e", "slhf"]].dropna()

print("Correlation:", d["slhf"].corr(d["e"]))

slope, b = np.polyfit(d["e"], d["slhf"], 1)
print("Slope:", slope, "Intercept:", b)
print("Residual std:", (d["slhf"] - (slope * d["e"] + b)).std())

sm = d.sample(min(10000, len(d)), random_state=0)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=sm, x="e", y="slhf", alpha=0.25, s=12)
plt.title("SLHF vs evaporation")
plt.show()
```

## precipitation zeros and near-zero values

```python
tp_mm = df["tp"] * 1000    # temporary Series for readability, no new column

print(tp_mm.describe())
print("Exact zeros:", (tp_mm == 0).sum(), f"({(tp_mm == 0).mean() * 100:.2f}%)")

for t in [0, 0.1, 1, 5, 10]:
    print(f"<= {t} mm: {(tp_mm <= t).mean() * 100:.2f}%")

plt.figure(figsize=(10, 4))
sns.histplot(np.log1p(tp_mm.clip(lower=0)), bins=60)
# plt.title("log(1 + precipitation in mm)")
plt.show()
```

### extended statistics

```python
cols = FEATURES + ["t2m"]

stats = df[cols].describe().T
stats["median"] = df[cols].median()
stats["variance"] = df[cols].var()
stats["skewness"] = df[cols].skew()
stats["kurtosis"] = df[cols].kurt()

print(stats.round(3))
```

## Exploration 3: Seasonality

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 7))
for ax, c in zip(axes.ravel(), ["t2m", "d2m", "ssr", "str", "tp", "e"]):
    df.groupby("month")[c].mean().plot(ax=ax, marker="o"); ax.set_title(f"Mean {c} by month")
plt.tight_layout(); plt.show()

plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x="month", y="t2m", fliersize=1); plt.show()
```

## Exploration 4: Year × month overview

```python
tab = df.pivot_table(index="year", columns="month", values="t2m", aggfunc="mean")
plt.figure(figsize=(10, 5))
sns.heatmap(tab, annot=True, fmt=".1f", cmap="coolwarm"); plt.show()
```

## exploration 5: yearly trend

```python
annual = df[df["year"] < 2026].groupby("year")["t2m"].mean()
slope, b = np.polyfit(annual.index, annual.values, 1)
plt.plot(annual.index, annual.values, "o-")
plt.plot(annual.index, slope * annual.index + b, "r--")
plt.title(f"trend = {slope*10:+.2f} °C / decade"); plt.show()
```

## Exploration 6: Geographic patterns

```python
m = df.groupby(["latitude", "longitude"])[["t2m", "tp", "sp", "ssr"]].mean().reset_index()
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, c, cm in zip(axes, ["t2m", "tp", "sp", "ssr"], ["coolwarm", "Blues", "viridis", "YlOrRd"]):
    sc = ax.scatter(m["longitude"], m["latitude"], c=m[c], cmap=cm, marker="s", s=25)
    plt.colorbar(sc, ax=ax); ax.set_title(c)
plt.show()
```

## Exploration 7: Variation across space or across time?

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
sns.histplot(df.groupby("month")["t2m"].mean(), ax=axes[0], bins=12)
axes[0].set_title("Mean t2m per month (time variation)")
sns.histplot(df.groupby("cell_id")["t2m"].mean(), ax=axes[1], bins=40)
axes[1].set_title("Mean t2m per location (space variation)")
plt.show()
```
